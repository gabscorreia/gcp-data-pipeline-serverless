```python
import pandas_gbq
import pandas as pd
from google.cloud import bigquery
from google.colab import auth

auth.authenticate_user()
```

```python
# 1. Configurações iniciais do projeto e do dataset bruto
project_id = "strong-planet-484414-a0"
dataset_id = "ecommerce_bronze"
location = "southamerica-east1"
```

```python
# Inicializando o cliente
client = bigquery.Client(project=project_id)

# 2. Listando automaticamente todas as tabelas presentes no dataset Bronze
tables = client.list_tables(dataset_id)

print(f"=== INICIANDO A INSPEÇÃO DE QUALIDADE===\n")

# 3. Criando um ciclo (loop) para percorrer tabela por tabela de forma automatizada
for table in tables:
  tabela_nome = table.table_id
  print(f"--------------------------------------------------")
  print(f"A inspecionar a tabela: {tabela_nome}")

  # 4. Lendo a tabela inteira do BigQuery para um DataFrame do Pandas
  query = f"SELECT * FROM `{project_id}.{dataset_id}.{tabela_nome}`"
  df = pandas_gbq.read_gbq(query, project_id=project_id, location=location)

  # 5. Extraindo métricas essenciais de qualidade de dados
  total_linhas = len(df)
  total_duplicadas = df.duplicated().sum()
  nulos_por_coluna = df.isnull().sum()

  # 6. Exibindo os resultados resumidos para esta tabela
  print(f"-> Total de linhas: {total_linhas}")
  print(f"-> Total de linhas duplicadas: {total_duplicadas}")
  print(f"-> Valores nulos por coluna: {nulos_por_coluna}")
  print(nulos_por_coluna[nulos_por_coluna > 0] if nulos_por_coluna.sum() > 0 else "   Nenhum valor nulo encontrado.")
  print()

print(f"=== INSPEÇÃO CONCLUÍDA COM SUCESSO ===")
```
